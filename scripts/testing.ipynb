```python
def read_d13C_atm_data(file):

    df = pd.read_table(str(file), sep = '\s+', header=None)
    df.rename(
        columns={
            0: 'year',
            1: 'd13C_atm',
            2: 'stnd_dev'
        }
    )

    df['year'] = df['year'].apply(lambda x: x*1000)

    d13C_atm_data = df.to_numpy()

    print(df.shape[1])

    return d13C_atm_data
    
    
    
```
